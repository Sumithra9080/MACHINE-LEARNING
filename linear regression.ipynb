```python
import pandas as pd
```

```python
d = pd.read_csv("Salary_Data.csv")
d
```

```python
d.info() #dataset information 
```

```python
d.columns[d.isnull().any()]
```

```python
import numpy as np
```

```python
d["Age"]=d["Age"].fillna(d["Age"].mean())
```

```python
d.columns[d.isnull().any()]
```

```python
d["Salary"]=d["Salary"].fillna(d["Salary"].mean())
```

```python
d.columns[d.isnull().any()]
```

```python
d["Years of Experience"]=d["Years of Experience"].fillna(d["Years of Experience"].mean())
```

```python
d["Gender"].value_counts()
```

```python
e=pd.get_dummies(d["Gender"])
e
```

```python
d=pd.concat([d,e],axis=1)
d
```

```python
d=d.drop(["Gender"],axis=1)  #remove the row and colaum
d
```

```python
d.info()
```

```python
#pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=d[["Years of Experience"]]
y=d["Salary"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.1,random_state=0)
x_train
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
y_pred
```

```python
x_test
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x_train,y_train)
plt.plot(x_test,y_pred,color="red")
plt.show()
```
